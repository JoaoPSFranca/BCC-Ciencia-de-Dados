O _dataset_ utilizado está disponível no [Kaggle](https://www.kaggle.com) e é a base [kc_house_data](https://www.kaggle.com/datasets/shivachandel/kc-house-data).

## Tratando as Datas

```r
# Verifica que não tem nada diferente de 000000 depois do T
unique(substr(kc$date,10,15))

# Acha onde está o 'e erro'
linha <- grep("e erro", kc$date)
kc[linha,]

# Tira a linha com erro
kc <- kc[-linha,]
```

```r
# Verifica os anos, meses e dias
unique(substr(kc$date,1,4))
unique(substr(kc$date,5,6))
unique(substr(kc$date,7,8))

# Criando as colunas de ano, mes e dia
kc$year <- as.integer(substr(kc$date,1,4))
kc$month <- as.integer(substr(kc$date,5,6))
kc$day <- as.integer(substr(kc$date,7,8))

# Transformando em date
kc$date <- as.Date(kc$date, tryFormats = c("%Y%m%d"))

# Sumary
print(summary(kc$date))
```

## IDs Duclicados

```r
# Ordena a base de dados por ID
kc <- kc[order(kc$id),]

# Pegar os duplicados
dup <- kc$id[duplicated(kc$id)]

# Itera sobre os duplicados para mostrar as informações
for (p in dup[1:10]) {
    print(kc[kc$id == p, c("id", "date")])
}

for (p in dup) {
    caso <- kc[kc$id == p,]
    caso <- caso[order(caso$date),]
    caso <- caso[nrow(caso),]
    caso <- kc[kc$id != p,]
    rbind(kc, caso)
}
```
