## 1. Visão Geral do Capítulo e Fundamentação Conceitual

### 1.1 A Pirâmide do Conhecimento (DIKW)
A Ciência de Dados transforma registros brutos em tomadas de decisão de alto impacto por meio de quatro estágios evolutivos:

1. **Dado (*Data*)**: É a unidade atômica quantificável ou categórica associada a uma entidade ou evento. Isoladamente, não possui significado intrínseco e não permite contexto (ex.: `1,78` ou `R$ 470.000,00` — sem contexto, não sabemos se é altura, preço, muito ou pouco).
2. **Informação (*Information*)**: É o dado devidamente processado, organizado, contextualizado e associado a um significado (ex.: *“A altura do candidato Lucas é 1,78m”* ou *“O preço médio do m² no bairro X é R$ 4.700,00”*).
3. **Conhecimento (*Knowledge*)**: É a informação interpretada, correlacionada e compreendida por especialistas ou algoritmos, permitindo modelar comportamentos e prever padrões futuros (ex.: *“Imóveis acima de 100m² neste bairro valorizam 12% ao ano”*).
4. **Ação / Decisão (*Wisdom / Action*)**: Implementação prática do conhecimento adquirido para otimizar processos de negócios, automatizar fluxos ou resolver problemas complexos.

### 1.2 Divisão de Funções em Times de Dados Modernos
* **Engenharia de Dados**: Foco na infraestrutura, ingestão massiva, levantamento, limpeza inicial e garantia da integridade e disponibilidade dos pipelines (ETL/ELT).
* **Business Intelligence (BI)**: Foco no diagnóstico do passado e presente através de dashboards, métricas de KPI e relatórios analíticos de negócio.
* **Ciência de Dados (Data Science)**: Foco na aplicação de ferramental estatístico e algoritmos de Machine Learning para prever cenários futuros (*Análise Preditiva e Prescritiva*).

## 2. A Era do Big Data e o Paradigma MapReduce

### 2.1 Limitações das Arquiteturas Tradicionais e os 5 V's
Bancos de dados relacionais clássicos (SQL centralizados) dos anos 1970/80 tornaram-se gargalos frente à explosão da Web e dispositivos conectados. Surgem os **5 V's do Big Data**:
* **Volume**: Ordens de grandeza em terabytes, petabytes e exabytes.
* **Velocidade**: Ingestão e processamento contínuo em tempo real (streaming).
* **Variedade**: Dados estruturados (tabelas), semiestruturados (JSON, XML) e não estruturados (textos, imagens, áudios).
* **Veracidade**: Filtragem de ruídos, fraudes e incertezas nos dados coletados.
* **Valor**: A capacidade de extrair retorno financeiro e inteligência estratégica a partir dos dados.

### 2.2 Arquiteturas de Dados: Centralizada vs. Replicada vs. Distribuída
| Arquitetura | Armazenamento | Filas e Latência | Tolerância a Falhas | Escalabilidade |
| :--- | :--- | :--- | :--- | :--- |
| **Centralizada** | Limitada a um servidor único | Fila única com gargalo de concorrência | Falha total caso o servidor caia | Vertical (*Scale-Up*, cara e com teto físico) |
| **Replicada** | Cópias completas em múltiplos nós | Requisições distribuídas entre réplicas | Tolerante a quedas parciais | Média para leituras, péssima para escritas massivas |
| **Distribuída** | Dados particionados em dezenas/centenas de nós | Filas locais paralelas sem gargalo único | Alta tolerância via replicação de blocos | Horizontal (*Scale-Out*, ilimitada e resiliente) |

### 2.3 O Paradigma MapReduce
Inspirado na programação funcional, o MapReduce divide um problema massivo em milhares de pequenas tarefas independentes executadas em paralelo:
1. **Map (Mapeamento)**: Cada nó processa uma fatia local dos dados e gera pares intermediários `(chave, valor)`.
2. **Shuffle / Sort (Agrupamento e Ordenação)**: O sistema distribui pela rede todos os registros que compartilham a mesma chave para o mesmo nó redutor.
3. **Reduce (Redução)**: Cada nó consolida, soma ou agrega os valores associados à sua chave, emitindo o resultado final consolidado.

### Modos de Operação e Principais Atalhos
O Jupyter possui dois modos de operação fundamentais:
* **Modo de Comando** (Borda esquerda **azul**, ativado pressionando `Esc`):
  * `A`: Insere uma célula acima (*Above*).
  * `B`: Insere uma célula abaixo (*Below*).
  * `M`: Converte a célula para **Markdown** (texto explicativo).
  * `Y`: Converte a célula para **Code** (código executável).
  * `D, D` (duas vezes): Deleta a célula selecionada.
  * `Z`: Desfaz a última exclusão de célula.
* **Modo de Edição** (Borda esquerda **verde**, ativado pressionando `Enter` em cima da célula):
  * `Shift + Enter`: Executa a célula atual e move para a próxima.
  * `Ctrl + Enter`: Executa a célula mantendo a seleção nela.

### Passo 2: Representando a Pirâmide DIKW em R

🎯 **O que fazer**:
Criar dados brutos não estruturados, transformá-los em informações organizadas e derivar conhecimento acionável via agregação estatística.

💻 **Como fazer**:

```r
# 1. Dados brutos (sem contexto intrínseco)
valores_brutos <- c(180, 220, 150, 480, 520, 190, 210, 490)

# 2. Informação (atribuindo semântica e contexto)
registro_lojas <- data.frame(
  filial = c("Loja A", "Loja A", "Loja A", "Loja B", "Loja B", "Loja C", "Loja C", "Loja B"),
  vendas_mil = valores_brutos,
  mes = rep(c("Jan", "Fev"), 4)
)
print("--- Informação Estruturada ---")
print(registro_lojas)

# 3. Conhecimento (modelagem e extração de padrões consolidados)
desempenho_filiais <- aggregate(vendas_mil ~ filial, data = registro_lojas, FUN = function(x) c(Total = sum(x), Media = mean(x)))
print("--- Conhecimento Derivado (Desempenho por Filial) ---")
print(desempenho_filiais)
```

💡 **Por que fazer**:
Ilustra na prática como dados isolados não geram valor comercial até serem contextualizados em estruturas relacionais e sumarizados para suportar o processo decisório.

### Passo 3: Implementando o Paradigma MapReduce em R

🎯 **O que fazer**:
Simular o algoritmo MapReduce para o problema clássico de Big Data: **identificar conexões mútuas (amigos em comum) em uma rede social distribuída**.

💻 **Como fazer**:

```r
# Rede social de exemplo: Usuário -> Lista de Amigos
rede_social <- list(
  Ana = c("Beto", "Caio", "Dora"),
  Beto = c("Ana", "Caio", "Edu"),
  Caio = c("Ana", "Beto", "Dora", "Edu"),
  Dora = c("Ana", "Caio"),
  Edu = c("Beto", "Caio")
)

# ETAPA 1: MAP (Gera pares chave-valor para cada par de amigos em potencial)
etapa_map <- function(rede) {
  pares_chave_valor <- list()
  for (usuario in names(rede)) {
    amigos <- sort(rede[[usuario]])
    if (length(amigos) >= 2) {
      for (i in 1:(length(amigos)-1)) {
        for (j in (i+1):length(amigos)) {
          chave <- paste(amigos[i], amigos[j], sep = "<->")
          pares_chave_valor[[length(pares_chave_valor) + 1]] <- list(chave = chave, valor = usuario)
        }
      }
    }
  }
  return(pares_chave_valor)
}

# ETAPA 2 & 3: SHUFFLE & REDUCE (Agrupa por chave e consolida os amigos em comum)
etapa_reduce <- function(pares_map) {
  tabela_agrupada <- list()
  for (item in pares_map) {
    ch <- item$chave
    val <- item$valor
    if (is.null(tabela_agrupada[[ch]])) {
      tabela_agrupada[[ch]] <- val
    } else {
      tabela_agrupada[[ch]] <- c(tabela_agrupada[[ch]], val)
    }
  }
  
  # Formatando para exibição clara
  resultado <- data.frame(
    Par_Usuarios = names(tabela_agrupada),
    Qtd_Amigos_Comum = sapply(tabela_agrupada, length),
    Amigos_Comum = sapply(tabela_agrupada, paste, collapse = ", ")
  )
  return(resultado)
}

# Executando o pipeline distribuído simulado:
pares_mapeados <- etapa_map(rede_social)
conexoes_comuns <- etapa_reduce(pares_mapeados)

cat("Total de pares emitidos na fase MAP:", length(pares_mapeados), "\n\n")
print(conexoes_comuns)
```

💡 **Por que fazer**:
Esse código reproduz fielmente a arquitetura de ferramentas como Hadoop e Apache Spark. Ao fatiar a computação em chaves independentes, o algoritmo pode ser distribuído para centenas de máquinas sem dependências de memória compartilhada, permitindo escalar horizontalmente para bilhões de conexões.
