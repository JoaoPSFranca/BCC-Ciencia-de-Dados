## 1. Visão Geral do Capítulo e Filosofia da Linguagem R

O R é uma linguagem de programação funcional e ambiente voltado à análise de dados, computação estatística e gráficos. Principais peculiaridades que todo cientista de dados deve dominar:
1. **Tudo é um objeto**: No R, variáveis, funções, gráficos e expressões são objetos em memória.
2. **Atribuição com `<-`**: Embora `=` funcione, a convenção universal é o operador de seta `<-` (lembre-se: `x <- 10` significa "atribua 10 à variável x").
3. **Indexação iniciada em 1**: Diferente de C, Java ou Python (que usam índice 0), o R adota a convenção matemática onde o primeiro elemento ocupa a **posição 1**.
4. **Vetorização nativa**: Operações matemáticas (soma, produto, comparações lógicas) são aplicadas a vetores inteiros de uma só vez, sem necessidade de laços `for` manuais.
5. **Regra de Reciclagem**: Ao operar dois vetores de comprimentos distintos, o R repete os elementos do menor até atingir o tamanho do maior. Se o comprimento do maior for múltiplo do menor, a operação ocorre sem avisos; caso contrário, é emitido um *warning*.

## 2. As Principais Estruturas de Dados do R

| Estrutura | Dimensões | Homogênea (mesmo tipo)? | Função de Criação | Exemplo de Aplicação |
| :--- | :--- | :--- | :--- | :--- |
| **Vetor Atômico** | 1D | Sim | `c()`, `:`, `seq()` | Séries de preços, idades, flags lógicas |
| **Fator (*Factor*)** | 1D | Sim (inteiros com labels) | `factor()` | Variáveis categóricas (sexo, escolaridade, estados) |
| **Matriz** | 2D | Sim | `matrix()` | Álgebra linear, transformações, imagens |
| **Data Frame** | 2D | Não (colunas heterogêneas) | `data.frame()` | Tabelas relacionais, bancos de dados |
| **Lista** | 1D recursiva | Não (qualquer tipo/tamanho) | `list()` | Saídas de modelos estatísticos complexos |

### Passo 1: Criação e Operações Vetorizadas Avançadas

🎯 **O que fazer**:
Construir vetores utilizando funções geradoras (`seq`, `rep`), aplicar operações aritméticas vetorizadas e compreender a regra de reciclagem.

💻 **Como fazer**:

```r
# Limpeza inicial do ambiente
rm(list = ls())

# 1. Criação de sequências regulares e repetições
seq_valores <- seq(from = 10, to = 50, by = 5)   # 10, 15, 20, ..., 50
rep_valores <- rep(c(1, 2, 3), times = 3)       # 1, 2, 3, 1, 2, 3, 1, 2, 3
rep_cada    <- rep(c("A", "B"), each = 4)       # A, A, A, A, B, B, B, B

cat("Sequência gerada:\n", seq_valores, "\n")
cat("Repetição agrupada:\n", rep_cada, "\n")

# 2. Operações vetorizadas diretas (sem laços for)
pesos_kg <- c(65, 80, 55, 95, 72)
alturas_m <- c(1.70, 1.82, 1.60, 1.90, 1.75)

# Cálculo instantâneo do IMC para todas as observações: IMC = peso / (altura^2)
imc <- pesos_kg / (alturas_m^2)
cat("\nValores de IMC calculados:\n", round(imc, 2), "\n")

# 3. Regra de Reciclagem (vetor de tamanho 5 operado com vetor de tamanho 1 ou submúltiplo)
taxa_adicional <- c(1.05) # acréscimo de 5%
pesos_com_taxa <- pesos_kg * taxa_adicional
cat("Pesos ajustados com reciclagem escalar:\n", pesos_com_taxa, "\n")
```

💡 **Por que fazer**:
A vetorização em R é implementada internamente em linguagem C e Fortran de alto desempenho. Evitar loops explícitos (`for`) reduz drasticamente o tempo de processamento em grandes volumes de dados e torna o código mais conciso e legível.

### Passo 2: Técnicas Avançadas de Fatiamento e Indexação

🎯 **O que fazer**:
Extrair fatias de dados utilizando as três formas de indexação do R: índices positivos, índices negativos (exclusão) e máscaras lógicas (filtragem booleana).

💻 **Como fazer**:

```r
idades <- c(18, 25, 33, 42, 16, 58, 29, 64, 15, 38)

# 1. Indexação Positiva: elementos específicos
primeiros_tres <- idades[1:3]
selecao_esparsa <- idades[c(2, 5, 8)]

# 2. Indexação Negativa: exclui os índices informados
sem_primeiro_e_ultimo <- idades[-c(1, length(idades))]

# 3. Indexação Lógica (Máscara Booleana)
# Passo 3a: criar o vetor de condições lógicas (TRUE / FALSE)
mascara_maiores <- idades >= 18 & idades < 60

# Passo 3b: filtrar apenas onde a máscara for TRUE
idades_adultos <- idades[mascara_maiores]

cat("Vetor completo:", idades, "\n")
cat("Primeiros três:", primeiros_tres, "\n")
cat("Sem primeiro e último:", sem_primeiro_e_ultimo, "\n")
cat("Apenas adultos (18 a 59 anos):", idades_adultos, "\n")
```

💡 **Por que fazer**:
A indexação lógica é o alicerce para qualquer filtragem em Ciência de Dados. Ao invés de iterar sobre linhas, expressamos regras declarativas que filtram dados instantaneamente com extrema fidelidade estatística.

### Passo 3: Construção e Álgebra de Matrizes

🎯 **O que fazer**:
Criar matrizes bidimensionais, configurar ordenação por linhas ou colunas, nomear dimensões e realizar operações de álgebra matricial.

💻 **Como fazer**:

```r
# Criando matriz 3x4 preenchida por coluna (padrão)
mat_col <- matrix(1:12, nrow = 3, ncol = 4)

# Criando matriz 3x4 preenchida por linha
mat_lin <- matrix(1:12, nrow = 3, ncol = 4, byrow = TRUE)

# Atribuindo rótulos legíveis às dimensões
rownames(mat_lin) <- c("Regiao_Sul", "Regiao_Sudeste", "Regiao_Norte")
colnames(mat_lin) <- c("Trim1", "Trim2", "Trim3", "Trim4")

print("Matriz preenchida por linha com rótulos:")
print(mat_lin)

# Fatiamento bidimensional: [linhas, colunas]
vendas_sudeste <- mat_lin["Regiao_Sudeste", ] # linha inteira
vendas_trimestre_4 <- mat_lin[, "Trim4"]      # coluna inteira

cat("\nVendas da Região Sudeste nos 4 trimestres:\n", vendas_sudeste, "\n")
cat("Vendas de todas as regiões no 4º trimestre:\n", vendas_trimestre_4, "\n")
```

💡 **Por que fazer**:
Matrizes são a representação padrão para cálculos numéricos puros (como matrizes de correlação, covariância, distâncias euclidianas e processamento de imagem), além de possuírem consumo de memória menor que DataFrames.

### Passo 4: Fatores e Tratamento de Variáveis Categóricas

🎯 **O que fazer**:
Representar dados categóricos nominais e ordinais utilizando `factors`, compreendendo como o R armazena níveis e mapeia valores internamente.

💻 **Como fazer**:

```r
# Vetor de strings com categorias ordinais (níveis hierárquicos)
respostas_pesquisa <- c("Médio", "Alto", "Baixo", "Médio", "Alto", "Alto", "Baixo")

# Fator Nominal (sem ordem intrínseca)
fator_nominal <- factor(respostas_pesquisa)
print("Fator nominal (ordem alfabética padrão dos níveis):")
print(fator_nominal)
print(levels(fator_nominal))

# Fator Ordinal (com ordem pré-definida de importância: Baixo < Médio < Alto)
fator_ordinal <- factor(
  respostas_pesquisa,
  levels = c("Baixo", "Médio", "Alto"),
  ordered = TRUE
)
print("\nFator ordinal estruturado:")
print(fator_ordinal)

# Frequência de cada categoria
print("\nTabela de contingência:")
print(table(fator_ordinal))
```

💡 **Por que fazer**:
Algoritmos de aprendizado supervisionado (como regressão linear e logística) exigem que variáveis categóricas sejam convertidas em variáveis *dummy* ou níveis codificados. Declarar explicitamente uma coluna como `factor` instrui o R a gerenciar os contrastes categóricos automaticamente.

### Passo 5: Estruturação e Enriquecimento de DataFrames

🎯 **O que fazer**:
Construir um DataFrame analítico consolidando múltiplos tipos, adicionar variáveis derivadas, inspecionar propriedades estruturais e realizar agregações por grupo.

💻 **Como fazer**:

```r
# Montagem de um DataFrame analítico
equipe <- data.frame(
  id = 101:106,
  nome = c("Lucas", "Beatriz", "Carlos", "Diana", "Eduardo", "Fernanda"),
  departamento = factor(c("TI", "Vendas", "TI", "RH", "Vendas", "TI")),
  horas_extras = c(12, 5, 20, 0, 8, 15),
  salario_base = c(6500, 4800, 7200, 4200, 5100, 6900),
  stringsAsFactors = FALSE
)

# Adicionando uma coluna calculada: valor da hora e salário total estimado
valor_hora <- equipe$salario_base / 160
equipe$remuneracao_total <- equipe$salario_base + (equipe$horas_extras * valor_hora * 1.5)

# Filtragem com a função subset
destaques_ti <- subset(equipe, departamento == "TI" & horas_extras > 10)

print("DataFrame consolidado:")
print(equipe[, c("nome", "departamento", "horas_extras", "remuneracao_total")])

print("\nProfissionais de TI com alta carga de horas extras:")
print(destaques_ti[, c("nome", "departamento", "horas_extras")])

# Média de remuneração total por departamento
media_por_depto <- aggregate(remuneracao_total ~ departamento, data = equipe, FUN = mean)
print("\nMédia salarial calculada por departamento:")
print(media_por_depto)
```

💡 **Por que fazer**:
O DataFrame é o coração da manipulação de dados em R. Saber combinar vetores, calcular novos campos derivados com base em regras de negócio e realizar agregações com `aggregate` é o fundamento necessário para qualquer análise estatística posterior.
