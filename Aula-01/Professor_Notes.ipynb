````r
nome <- c("Abel","Bianca","Caio","Dênis","Ema", "Fabio", "Gus", "Helena")
nota <- c(   6.5,     9.3,   2.2,    6.9,  7.0,     8.4,   7.1,      7.9)
freq <- c(  0.68,    0.81,  0.57,   0.71, 0.83,    0.90,  0.85,     0.78)
sexo <- c(   'M',     'F',   'M',    'M',  'F',     'M',   'X',      'F')
````

````r
nome
````

````r
sexo
````

````r
classe <- data.frame(nome,nota,freq,sexo)
classe
````

````r
classe[1,4]
````

````r
classe[2,3]
````

````r
classe[2:5,3]
````

````r
classe[c(2,5,7),c(1,3,4)]
````

````r
classe[-3,]
````

````r
class(classe)
````

````r
classe[classe$nota>7 & classe$freq>0.8,1]
````

````r
mean(classe[classe$nota>7 & classe$freq>0.8,2])
````

Listar os nomes dos alunos com frequência maior que 80%

````r
classe[classe$freq>0.8, "nome"]
````

Listar alunos aprovados (nota >= 6.0 e freq >= 75%)

````r
aprovados <- classe$nota >= 6.0 & classe$freq >= 0.75
classe[aprovados,]
````

Listar nota e frequência dos alunos reprovados por nota mas aprovados por frequência

````r
reprovados_freq <- classe$nota>=6.0 & classe$freq<0.75
classe[reprovados_freq, c("nota", "freq")]
````
