### 3 - vetor

O vetor `temp` a seguir apresenta a temperatura °C de 50 pessoas após 20 minutos de terem sido inoculadas com um medicamento experimental:

```r
set.seed(997)
temp <- round(rnorm(50, 38.15, 1), 1)
print(temp)
```

Diversos procedimentos de análise de dados necessitam trabalhar com as variáveis padronizadas, para que a grandeza de uma variável com valores muito grandes (área em m², por exemplo) não prejudique outra com valores muito pequenos (concentração de poluentes no ar). Assim, há dois métodos de padronização bastante utilizados: o método ***Min_Max***:  

$$ dados\_min\_max = \frac{dados_i - min(dados)} {max(dados) - min(dados)} $$  
Assim, se _dados_ for {1, 2, 3}, _dados_min_max_ será {0, 0,5 1} ou seja, cada posição menos o mínimo dividida pela amplitude.

__a)__ Crie e imprima a variável `temp_min_max`, que é a variável `temp` padronizada por Min_Max.  

```r
temp_min_max <- (temp - min(temp)) / (max(temp) - min(temp))
print(temp_min_max)
```

```r
porc <- c(0.4, 0.17, 0.008, 0.53)
area <- c(2300, 2800, 5400, 1920)
```

```r
porc_min <- min(porc)
porc_max <- max(porc)
porc_min
porc_max
```

```r
porcmm <- (porc - porc_min) / (porc_max - porc_min)
print(porcmm)
```

```r
area_min <- min(area)
area_max <- max(area)
area_min
area_max
area_max - area_min  # amplitude é o maior valor - menor valor
```

```r
areamm <- (area - area_min) / (area_max - area_min)
print(areamm)
```

```r
print(porcmm)
print(areamm)
```

Outra técnica é padronização ***Znorm***:  
$$ dados\_znorm = \frac{dados_i - média} {desvio} $$  
Sendo que _média_ e _desvio_ são:  
$$ média =  \frac {\sum dados_i} {n} , \hspace{2cm} desvio = \frac {\sum {(dados_i - média)^2}} {n-1} $$
É claro que <span style="font-family:Serif; font-size:120%">R</span> possui funções para média e desvio:

```r
media <- mean(temp)
media
```

```r
desvio <- sd(temp)
desvio
```

__b)__ Crie e imprima a variável `temp_znorm`, que é a variável `temp` padronizada por ZNorm.

```r
mean(temp)
temp_znorm <- (temp - mean(temp)) / sd(temp)
print(temp_znorm)
```
