1 - Crie um vetor `v` de 10 posições numéricas

```python
v <- c(1, 7, 9, 10, 15, 24, 83, 92, 5, 0)
v
```

2 - Salve as posições 2, 3, 5 e 9 de `v` em um novo vetor `s`.

```python
s <- v[c(2, 3, 5, 9)]
s
```

3 - Obtenha um vetor `t` com apenas os valores maiores que 10 de `v`.

```python
t <- v[v>10]
t
```

4 - Obtenha um vetor `u` sem as posições 1, 2, 3, 8, 9 e 10 de `v`.

```python
u <- v[-c(1, 2, 3, 8, 9, 10)]
u
```
